# Notebook 0: Dataset structuur
Het doel van dit notebook is om de structuur van de dataset helder te krijgen.

```python
from pathlib import Path

DATA_DIR = Path.cwd().parent / "datasets iecon" / "datasets iecon"
```

```python
def print_folder_tree(path, niveau=0):
    for folder in sorted(path.iterdir()):
        if folder.is_dir():
            zip_count = len(list(folder.rglob("*.zip")))

            print("    " * niveau + folder.name + f" [{zip_count} ZIP-bestanden]")

            print_folder_tree(folder, niveau + 1)
```

```python
print(DATA_DIR.name)
print_folder_tree(DATA_DIR)
```
